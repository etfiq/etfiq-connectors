# ETFIQ

[ETFIQ](https://etfiq.com) publishes free data on US-listed ETFs: fees, size, returns, holdings, fund flows and each fund's return against its own benchmark. `langchain-etfiq` wraps it as six tools. No API key.

Source: ETFIQ (etfiq.com). Every record carries its `as_of` date, `source_url` and an `attribution` line.

## Overview

| Class | Package | Serializable | JS support |
|---|---|---|---|
| `EtfiqToolkit` | `langchain-etfiq` | no | no |

### Tool features

Read-only, keyless. Tools: `etfiq_search`, `etfiq_profile`, `etfiq_holdings`, `etfiq_flows`, `etfiq_compare`, `etfiq_screen`.

## Instantiation

```python
from langchain_etfiq import EtfiqToolkit

toolkit = EtfiqToolkit()
tools = toolkit.get_tools()
[t.name for t in tools]
```

## Invocation

```python
by_name = {t.name: t for t in tools}
by_name["etfiq_search"].invoke({"query": "covered call", "limit": 3})
```

```python
by_name["etfiq_profile"].invoke({"ticker": "JEPI"})
```

```python
by_name["etfiq_flows"].invoke({"ticker": "SPY"})[:2]
```

```python
by_name["etfiq_compare"].invoke({"tickers": ["SPY", "VOO"]})
```

```python
by_name["etfiq_screen"].invoke({"category": "Income", "max_fee": 0.5, "sort": "yield_12m", "limit": 5})
```

## Use within an agent

```python
from langchain.agents import create_agent

agent = create_agent("anthropic:claude-sonnet-4-5", tools)
result = agent.invoke({"messages": [("user", "How did JEPI do against its benchmark over the last year? Cite the date.")]})
print(result["messages"][-1].content)
```

## Using the MCP server instead

ETFIQ's MCP server has 33 tools, including committee reviews, scorecards and model portfolios.

```python
from langchain_mcp_adapters.client import MultiServerMCPClient

client = MultiServerMCPClient({"etfiq": {"url": "https://mcp.etfiq.com/mcp", "transport": "streamable_http"}})
mcp_tools = await client.get_tools()
len(mcp_tools)
```
