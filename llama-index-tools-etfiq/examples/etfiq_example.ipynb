# ETFIQ Tool Spec

Free ETF data from [ETFIQ](https://etfiq.com): fees, size, returns, holdings, flows, benchmark gap. No API key. Source: ETFIQ (etfiq.com).

```python
from llama_index.tools.etfiq import EtfiqToolSpec

spec = EtfiqToolSpec()
print(spec.profile("JEPI"))
```

```python
from llama_index.core.agent.workflow import FunctionAgent
from llama_index.llms.openai import OpenAI

agent = FunctionAgent(tools=spec.to_tool_list(), llm=OpenAI(model="gpt-4.1"))
print(await agent.run("Which income ETFs under 0.5% fee have the highest yield? Give the date of the data."))
```
